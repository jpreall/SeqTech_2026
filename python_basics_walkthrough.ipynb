## Object Oriented Programming Languages

As a beginner, you will mostly be interacting with other people's code. You'll probably load your data in and run some pre-baked analyses using tutorials as a guide. You'll likely be loading your data into **objects** defined by the package you are using.  

An **object** is a bundle of data and functions that belong together.  
It keeps related information (attributes) and the operations you can do on that information (methods) in one self-contained unit.  
	•	**Attributes** are the labeled contents inside (e.g. `cell_type`, `species`, `counts`). You access them with a dot: `adata.obs`  
	•	**Methods** are the standard actions you can perform on it (e.g., normalize(), plot(), subset()).

## Class ≈ Object
*class* and *object* might get used interchangeably.  They're essentially the same thing. Pedantically, **class** is the *function* used to create a new **object**


````python
class Person:
    def __init__(self, name, sex, profession):
        self.name = name
        self.sex = sex
        self.profession = profession
        self.vibe = 'mysterious but confident'

    def work(self):
        print(f"{self.name} is working as a {self.profession}.")

    def study(self, topic):
        print(f"{self.name} is studying {topic}.")
        
    def __repr__(self):
        return f"{self.name} is a {self.sex} {self.profession} who is {self.vibe}"
````

````python
jon = Person('Jon','male','ninja')
jon
````

## 5. Collections: list, tuple, set, dict  
Items are separeted by `,`
- **list**: `[ ]` ordered, mutable
- **dict**: `{key:value}` "Card catalog" -> rapid retrieval of stashed data or variables
    - kind of like an **class** but easier to build
    - jon = {'name':'jon', 'age':44, 'jobs':['scientist','dad']}
- **set**: `{ }` unique items, no order
- **tuple**: `( )` ordered, immutable. Often used for things like (x,y) coordinates

## 7. Functions
Define with `def`. Return values with `return`. Use docstrings for quick help.

````python
def mean(values):
    """Return the arithmetic mean of a non-empty iterable of numbers."""
    total = 0
    n = 0
    for v in values:
        total += v
        n += 1
    return total / n

mean([1, 2, 3, 4])
````

### Built-in functions
Python has a lot of useful built-in functions for every day use:  
  * `range()` — efficient iteration over numeric sequences (looping over indices, chunking, windowing).
  * `len()` — get the size of sequences, arrays, DataFrames, FASTA entries, etc.
  * `enumerate()` — iterate with both index and value (great for parsing and scanning lists).
  * `zip()` — combine multiple sequences in parallel (useful for pairing IDs, coordinates, sequences).
  * `sum()` — quick aggregation of numeric lists or boolean masks.
  * `sorted()` — order items or records, often with custom keys.
  * `min()` / `max()` — get extrema in numeric or sequence data.
  * `set()` — fast membership testing and deduplication (gene lists, ID sets).
  * `any()` / `all()` — concise boolean logic across lists or arrays.

## 8. Comprehensions
Compact ways to build lists/dicts/sets

````python
squares = [i*i for i in range(6)]
evens = {i for i in range(10) if i % 2 == 0}
lengths = {w: len(w) for w in ["alpha", "beta", "gamma"]}
````
